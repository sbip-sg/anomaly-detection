```python
import json
```

```python
def detectbalance(txdict):
    txs = 0
    for key in list(txdict.keys()):
        possiblelist = []
        for flinfo in txdict[key]:
            flassets = flinfo['assets']
            possible = False
            signs = []
            for asset in flassets:
                sign = asset['sign']
                if sign not in signs:
                    signs.append(sign)
            if len(signs) == 1:
                possible = signs[0]
            possiblelist.append(possible)
        possibility = 0
        if len(possiblelist) != 0:
            possibility = sum(possiblelist)/len(possiblelist)
        if possibility > 0.5:
            txs += 1
    return txs
```

```python
with open('attack_balances.json') as inputs:
    attack = json.load(inputs)
```

```python
detectbalance(attack)
```
