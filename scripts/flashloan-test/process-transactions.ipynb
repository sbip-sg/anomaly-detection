```python
def getflashloantraces(trace):
    data_map = trace['dataMap']
    ids = [int(id) for id in data_map.keys()]
    ids = sorted(ids)
    functions = []

    for id in ids:
        t = data_map[str(id)]
        if 'invocation' in t:
            if "decodedMethod" in t['invocation'] and  t['invocation']["decodedMethod"]:
                if "name" in t['invocation']["decodedMethod"]:
                    # Collect all transaction traces with name as flashloan
                    if t['invocation']["decodedMethod"]['name'].lower() == 'flashloan':
                        functions.append(t['invocation']["decodedMethod"])
    return functions
```
